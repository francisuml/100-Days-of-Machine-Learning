```python
#Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.decomposition import PCA
import plotly.express as px
```

```python
#Load Dataset
url = "https://raw.githubusercontent.com/dD2405/Twitter_Sentiment_Analysis/master/train.csv"
df = pd.read_csv(url)
```

```python
#Ensure that there is no missing values and no duplicates
df = df.dropna().drop_duplicates().reset_index(drop=True)
```

## Text Preprocessing
Clean and prepare the text for clustering, topic modeling, or vectorization.
* We are going to remove URLs, Strip out Twitter-line mentions and hashtags, Removes punctuations
* We are going to use stopwords to remove common English stopwords
* Apply Stemming to reduce words to their base form
Clean and normalized input improves results.

```python
nltk.download('stopwords')

def clean_text(text):
    text = re.sub(r'http\S+|www\S+|http\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'\@w+|\#', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = text.lower()
    text = " ".join([PorterStemmer().stem(word) for word in text.split() if word not in stopwords.words('english')])
    return text
```

```python
df['clean_text'] = df['tweet'].apply(clean_text)
```

## Vectorization with TF-IDF

* Instantiate the model
* Fit the model

```python
# Instantiate and Fit the model
vectorizer = TfidfVectorizer(max_features=1000)
X = vectorizer.fit_transform(df['clean_text'])
```

##  K-Means Clustering

```python
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans_labels=kmeans.fit_predict(X)
```

```python
df['KMeans_Cluster'] = kmeans_labels
```

```python
df
```

```python
df['KMeans_Cluster']
```

## Hierarchical Clustering


```python
hierarchical = AgglomerativeClustering(n_clusters=5)
hierarchical_labels = hierarchical.fit_predict(X.toarray())
df['Hierarchical_Cluster'] = hierarchical_labels
```

## Visualization with plotly


```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_pca['Cluster'] = df['KMeans_Cluster'].astype(str)

fig = px.scatter(df_pca, x='PC1', y='PC2', color='Cluster',
                 title='K-Means Clustering on Text Data',
                 labels={'Cluster': 'Cluster ID'})

fig.show()
```

The plot visualizes the result of K-Means clustering applied to text data that has been reduced to 2 dimensions using PCA (Principal Component Analysis).
* Each dot represents a tweet/document.
* The x-axis and y-axis are PC1 and PC2, the top 2 principal components that capture the most variance in the data.
* Colors represent different clusters assigned by the K-Means algorithm (Cluster 0, 1, 2, 3, 4).

### Insights:
* `Clear Groupings`:
* Clusters are well-separated, especially Cluster 0 (blue) and Cluster 2 (purple), suggesting distinct groups of text documents.
* `Overlap in Clusters`:
* Some clusters (e.g., orange and purple) are slightly overlapping, indicating that a similar topics or semantic on its content.
* `Dense Central Region`:
* Many points are tightly packed around the origin, which is an implication of high similarity or a redundancy in certain tweets.
* Cluster 4 (green) is sparse, indicating that it might be an `outlier group` or represent a rare category/topic.

This plot confirms that K-Means clustering, even reduced to 2D using Principal Component Analysis, can uncover distinct groupings of text data. 
The clustering helps:
* Group similar documents together.
* Identify dominant topics or themes in a corpus.
* Enable downstream tasks like topic labeling or content recommendation.
However, some overlap between clusters suggests that a deeper or more nuanced model (e.g., LDA or embeddings) might yield better separation.<br>
**PCA reduction is useful for visualization but may lose some semantic detail from high-dimensional space.**
